```python
padding_context = [0, 100, 1_000, 10_000, 50_000]
benign_score = [72.0, 69.2, 67.2, 40, 20]
```

```python
import matplotlib.pyplot as plt


# Plot
plt.figure(figsize=(8, 5))
plt.plot(padding_context, benign_score, marker='o', linestyle='-', linewidth=2)
plt.xlabel('Padding Context Length')
plt.xscale("log")
plt.ylabel('Benign Score')
plt.title('Benign Score vs. Padding Context Length')
plt.grid(True)
plt.tight_layout()
plt.show()
```
