### Back-of-the-hand cost calculation for input tokens expenditure

* Models are `GPT-4o-mini`, `GPT4.1-nano`, `GPT5-nano`

```python
# Token usage per evaluation run
# 44 input samples with 6 different context lengths
# The values are for paddings = 0, 100, 1000, 2000, 5000, 10000
tokens = 153_000 + 198_870 + 388_454 + 551_915 + 1_291_944 + 2_468_259
```

```python
print(f"Total number of input tokens per evaluation run: {tokens}")
```

#### Cost as per the OAI website

```python
o4_mini_cost = 0.60
nano4_cost = 0.20
nano5_cost = 0.050

o4_mini_cache = 0.30
nano4_cache = 0.05
nano5_cache = 0.005
```

#### Cost per evaluation run
Based on logs observation, around half of the tokens are cached

```python
per_run_total_cost_4o_mini = tokens * o4_mini_cost / (
    1_000_000 * 2
) + tokens * o4_mini_cache / (1_000_000 * 2)  # around half of the tokens are cached
print(f"GPT4o-mini per evaluation run cost: $ {per_run_total_cost_4o_mini:.3f}")
```

```python
per_run_total_cost_41_nano = tokens * nano4_cost / (
    1_000_000 * 2
) + tokens * nano4_cache / (1_000_000 * 2)  # around half of the tokens are cached
print(f"GPT4.1-nano per evaluation run cost: $ {per_run_total_cost_41_nano:.3f}")
```

```python
per_run_total_cost_5_nano = tokens * nano5_cost / (
    1000000 * 2
) + tokens * nano5_cache / (1000000 * 2)  # around half of the tokens are cached
print(f"GPT5-nano per evaluation run cost: $ {per_run_total_cost_5_nano:.3f}")
```

```python
seeds = 3  # 7, 21, 42
padding_type = 3  # before, after, around the task
task_types = 2  # benign and harmful

total_cost = (
    (
        per_run_total_cost_41_nano
        + per_run_total_cost_4o_mini
        + per_run_total_cost_5_nano
    )
    * seeds
    * padding_type
    * task_types
)
print(f"Total input token cost of the experiments: $ {total_cost:.3f}")
```

```python
## LLM judge cost
## Used GPT 4o mini

total_judgin_tokens = 6_480_000
total_judgin_cost = o4_mini_cost * total_judgin_tokens / (
    2 * 1_000_000
) + o4_mini_cache * total_judgin_tokens / (2 * 1_000_000)
print(f"LLM judge input token cost: $ {total_judgin_cost:.3f}")
```

```python
total = total_cost + total_judgin_cost
print(f"Total input token cost: $ {total:.3f}")
```
